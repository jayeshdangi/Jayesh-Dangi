```python
array=np.linspace(12,22,num=15,dtype=int)
```

```python

def variance_std(*args):
    total=0
    for i in args:
        total+=(i-np.mean(args))**2
    calculated_variance=total/len((args))
    return calculated_variance ,calculated_variance**(1/2)
```

```python
variance_std(*array)
```

```python

def variance_sample_std(*args):
    total=0
    for i in args:
        total+=(i-np.mean(args))**2
    variance_sample_std=total/len((args)-1)
    return variance_sample_std ,variance_sample_std**(1/2)
```
