```python
from sklearn.preprocessing import  PowerTransformer
```

```python
# for yeo-jonhson method using  negative value
X=np.array([
    [-1],
    [0],
    [3],
    [5],
    [10],
    [50]
])
pt=PowerTransformer(method='yeo-johnson')
X_transformed=pt.fit_transform(X)
print('original:')
print(X.ravel())
print('\nTransformed:')
print(np.round(X_transformed.ravel(),3))
print('nLearned lambda:')
print(pt.lambdas_)
```

```python
# for Box-cox method using  positive value
y=np.array([[1],[1],[3],[5],[10],[50]])
# create the transformed
pt=PowerTransformer(method='box-cox')
# fit and transformed
y_transformed=pt.fit_transform(y)
print('original:')
print(y.ravel())
print('\nTransformed:')
print(np.round(y_transformed.ravel(),3))
print('nLearned lambda:')
print(pt.lambdas_)
```

```python
import pandas as pd
import numpy as np

data = {
    'Name': ['Amit', 'Priya', 'Rahul', 'Sneha', 'Vikas'],
    'Gender': ['Male', 'Female', np.nan, 'Female', 'Male'],          # Categorical with Null
    'City': ['Delhi', 'Mumbai', 'Delhi', np.nan, 'Bangalore'],       # Categorical with Null
    'Age': [25, np.nan, 30, 22, np.nan],                            # Numerical with Null
    'Salary': [50000, 60000, np.nan, 45000, 70000],
    'department':['HR','NaN','Finance','IT','IT']               # Numerical with Null
}

df = pd.DataFrame(data)
df
```

```python
from sklearn.impute import SimpleImputer
```

```python
imputer=SimpleImputer(strategy='mean')
imputer.fit_transform(df[['Salary']])
```

```python
imputer=SimpleImputer(strategy='mean')
df['new_salary']=imputer.fit_transform(df[['Salary']]).flatten()
```

```python
df
```

```python
imputer=SimpleImputer(strategy='median')
df['median_salary']=imputer.fit_transform(df[['Salary']]).flatten()
```

```python
df
```

```python
imputer=SimpleImputer(strategy='most_frequent',)
df['new_cityt']=imputer.fit_transform(df[['City']]).flatten()
```
