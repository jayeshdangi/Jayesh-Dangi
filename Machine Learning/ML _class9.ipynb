```python
# Step 1: Import libraries
import pandas as pd
import numpy as np

from statsmodels.stats.outliers_influence import (
    variance_inflation_factor
)

# Step 2: Create dataset
df = pd.DataFrame({
    "Area_sqft": [
        1000, 1500, 2000, 2500, 3000, 3500
    ],
   # "Area_sqm": [
     #   92.9, 139.4, 185.8, 232.3, 278.7, 325.2
    #],
    #"Bedrooms": [2, 3, 3, 4, 4, 5],
    "Age": [15, 10, 8, 5, 3, 2]
})



# Step 4: Calculate VIF
X = df.copy()

vif_data = pd.DataFrame()
vif_data["Feature"] = X.columns

vif_data["VIF"] = [
    variance_inflation_factor(
        X.values, i
    )
    for i in range(X.shape[1])
]

# Step 5: Display VIF
print("\nVIF Results:")
vif_data.round(2)
```

so the variannce is highly spread throughout the data
