```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix,classification_report
from sklearn.model_selection import train_test_split
```

```python
sns.load_dataset('tips')
```

```python
df=sns.load_dataset('tips')
```

```python
df['sex']=df['sex'].map({'Male':1,'Female':0})
```

```python
df['smoker']=df['smoker'].map({'Yes':1,'No':0})
```

```python
df
```

```python
df.columns
```

```python
x=df[['total_bill','tip','smoker','size']]
y=df['sex']
```

```python
x_train, x_test, y_train, y_test=train_test_split(
    x,y,test_size=0.2,random_state=50
)
```

```python
log_res=LogisticRegression()
```

```python
x
```

```python
log_res.fit(x_train,y_train)
```

```python

y_pred=log_res.predict(x_test)
y_pred
```

```python
confusion_matrix(y_test,y_pred)
```

```python
print(classification_report(y_pred,y_test))
```

accuracy  is the summation of the all element in the coinfusion matres

```python
threshold=np.arange(0,1,0.1)
```

```python
from sklearn.metrics import recall_score,precision_score,f1_score
```

```python
result=[]
for vth in threshold:
    y_pred=(y_pred>vth).astype(int)
    precision_score_1=precision_score(y_test,y_pred)
    recall_score_1=recall_score(y_test,y_pred)
    f1_score_1=f1_score(y_test,y_pred,zero_division=0)
    result.append([vth,precision_score_1,recall_score_1,f1_score_1])
```

```python
result
```

```python
recall_score(y_pred,y_test)
```

```python
df=pd.DataFrame(result,columns=['vth','precision','recall','f1_score'])
```
