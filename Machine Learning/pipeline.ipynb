```python
# ==========================================
# STEP 1: IMPORT LIBRARIES
# ==========================================

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LinearRegression

from sklearn.metrics import mean_absolute_error, r2_score


# ==========================================
# STEP 2: LOAD RAW DATASET
# ==========================================

df = pd.read_csv(r"C:\Users\Jayesh Dangi\OneDrive\Desktop\machine learning\Machine Learning\student_marks .csv")

print("\nRAW DATASET")
print(df.head())

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nCOLUMN NAMES")
print(df.columns)


# ==========================================
# STEP 3: SEPARATE X AND Y
# ==========================================

X = df.drop("marks", axis=1)

y = df["marks"]

print("\nFEATURES (X)")
print(X.head())

print("\nTARGET (Y)")
print(y.head())


# ==========================================
# STEP 4: TRAIN TEST SPLIT
# ==========================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

print("\nTRAINING DATA SHAPE")
print(X_train.shape)

print("\nTESTING DATA SHAPE")
print(X_test.shape)


# ==========================================
# STEP 5: NUMERICAL PIPELINE
# ==========================================

num_features = [
    "hours_studied",
    "attendance"
]

num_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="mean")
    ),

    (
        "scaler",
        StandardScaler()
    )
])


# ==========================================
# STEP 6: COLUMN TRANSFORMER
# ==========================================

preprocessor = ColumnTransformer([
    (
        "numerical",
        num_pipeline,
        num_features
    )
])


# ==========================================
# STEP 7: COMPLETE ML PIPELINE
# ==========================================

model_pipeline = Pipeline([
    (
        "preprocessor",
        preprocessor
    ),

    (
        "model",
        LinearRegression()
    )
])


# ==========================================
# STEP 8: TRAIN THE MODEL
# ==========================================

model_pipeline.fit(X_train, y_train)

print("\nMODEL TRAINED SUCCESSFULLY")


# ==========================================
# STEP 9: MAKE PREDICTIONS
# ==========================================

y_pred = model_pipeline.predict(X_test)

print("\nPREDICTED MARKS")

print(y_pred)


# ==========================================
# STEP 10: MODEL EVALUATION
# ==========================================

mae = mean_absolute_error(y_test, y_pred)

r2 = r2_score(y_test, y_pred)

print("\nMODEL EVALUATION")

print("Mean Absolute Error:", round(mae, 2))

print("R2 Score:", round(r2, 2))


# ==========================================
# STEP 11: ACTUAL VS PREDICTED MARKS
# ==========================================

comparison = pd.DataFrame({
    "Actual_Marks": y_test.values,

    "Predicted_Marks": y_pred.round(2),

    "Error": (y_test.values - y_pred).round(2)
})

print("\nACTUAL VS PREDICTED MARKS")

print(comparison)


# ==========================================
# STEP 12: PREDICT FOR A NEW STUDENT
# ==========================================

new_student = pd.DataFrame({
    "hours_studied": [7],

    "attendance": [90]
})

new_prediction = model_pipeline.predict(new_student)

print("\nNEW STUDENT PREDICTION")

print("Predicted Marks:", round(new_prediction[0], 2))
```
