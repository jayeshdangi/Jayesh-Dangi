```python
#Question 1: Comparing Mean, Median, and Mode Imputation
#A. Identify the missing values in every column. Apply mean imputation and median imputation separately to the numerical columns, and mode imputation to the categorical columns. Create separate completed datasets and compare the imputed values, especially for Age and Monthly_Income, where extreme values may affect the mean.
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

df = pd.DataFrame({
"Age": [22, 25, np.nan, 30, 35, 100, np.nan, 28],
"Monthly_Income": [20000, 25000, 30000, np.nan,
35000, 500000, 28000, np.nan],
"City": ["Delhi", "Mumbai", "Delhi", np.nan,
"Pune", "Delhi", np.nan, "Mumbai"],
"Purchase_Category": ["Electronics", "Clothing", np.nan,
"Electronics", "Clothing", "Electronics",
"Clothing", np.nan]
})

numerical_cols = ["Age", "Monthly_Income"]
categorical_cols = ["City", "Purchase_Category"]

# Identify missing values

print("Missing values in each column:")
print(df.isnull().sum())

# Mean imputation for numerical columns

mean_imputer = SimpleImputer(strategy="mean")
df_mean = df.copy()

df_mean[numerical_cols] = mean_imputer.fit_transform(
df[numerical_cols]
)

# Median imputation for numerical columns

median_imputer = SimpleImputer(strategy="median")
df_median = df.copy()

df_median[numerical_cols] = median_imputer.fit_transform(
df[numerical_cols]
)

# Mode imputation for categorical columns

mode_imputer = SimpleImputer(strategy="most_frequent")

df_mean[categorical_cols] = mode_imputer.fit_transform(
df[categorical_cols]
)

df_median[categorical_cols] = mode_imputer.fit_transform(
df[categorical_cols]
)

print("\nDataset with Mean and Mode Imputation:")
print(df_mean)

print("\nDataset with Median and Mode Imputation:")
print(df_median)

print("\nMean Imputation Values:")
print(pd.Series(mean_imputer.statistics_, index=numerical_cols))

print("\nMedian Imputation Values:")
print(pd.Series(median_imputer.statistics_, index=numerical_cols))

print("\nCategorical Mode Values:")
print(pd.Series(mode_imputer.statistics_, index=categorical_cols))
```

```python
#B. Calculate the mean, median, and standard deviation of the numerical columns before and after each imputation strategy. Explain how the choice of strategy changes the statistical characteristics of the data and justify which strategy is more appropriate for each column.
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

df = pd.DataFrame({
"Age": [22, 25, np.nan, 30, 35, 100, np.nan, 28],
"Monthly_Income": [20000, 25000, 30000, np.nan,
35000, 500000, 28000, np.nan]
})

# Calculate statistics before imputation

print("Original Statistics:")
print(df.describe().loc[["mean", "50%", "std"]])

# Apply mean imputation

mean_imputer = SimpleImputer(strategy="mean")
df_mean = pd.DataFrame(
mean_imputer.fit_transform(df),
columns=df.columns
)

# Apply median imputation

median_imputer = SimpleImputer(strategy="median")
df_median = pd.DataFrame(
median_imputer.fit_transform(df),
columns=df.columns
)

# Compare statistics

for name, data in [
("Mean Imputation", df_mean),
("Median Imputation", df_median)
]:
    print(f"\n{name}")
    print("Mean:")
    print(data.mean())
    print("\nMedian:")
    print(data.median())
    print("\nStandard Deviation:")
    print(data.std())
```

```python
#Question 2: Implementing SimpleImputer with Different Strategies
#A. Use SimpleImputer to create a preprocessing workflow that applies mean to Patient_Age, median to Blood_Sugar, and most_frequent to both categorical columns. Display the completed DataFrame with its original column names and verify that no missing values remain.
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer

df = pd.DataFrame({
"Patient_Age": [25, 32, np.nan, 45, 52, np.nan, 38, 60],
"Blood_Sugar": [90, 110, 105, np.nan, 250, 130, np.nan, 115],
"Blood_Group": ["A", "B", "O", np.nan, "A", "B", np.nan, "A"],
"Treatment": ["Basic", np.nan, "Advanced", "Basic", "Advanced", "Basic", np.nan, "Basic"]
})

preprocessor = ColumnTransformer(
transformers=[
("age", SimpleImputer(strategy="mean"), ["Patient_Age"]),
("sugar", SimpleImputer(strategy="median"), ["Blood_Sugar"]),
("group", SimpleImputer(strategy="most_frequent"), ["Blood_Group"]),
("treatment", SimpleImputer(strategy="most_frequent"), ["Treatment"])
],
verbose_feature_names_out=False
)

result = preprocessor.fit_transform(df)

df_completed = pd.DataFrame(
result,
columns=preprocessor.get_feature_names_out()
)

# Restore the original column order
df_completed = df_completed[df.columns]

print("Completed Dataset:")
print(df_completed)

print("\nMissing Values:")
print(df_completed.isnull().sum())

print("\nAny Missing Values?")
print(df_completed.isnull().values.any())
```

```python
#B. Create a second workflow that uses strategy="constant" to replace missing categorical values with "Unknown" and missing numerical values with 0. Compare the results of both workflows and discuss why replacing missing blood sugar measurements with zero may lead to misleading conclusions
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer

df = pd.DataFrame({
"Patient_Age": [25, 32, np.nan, 45, 52, np.nan, 38, 60],
"Blood_Sugar": [90, 110, 105, np.nan, 250, 130, np.nan, 115],
"Blood_Group": ["A", "B", "O", np.nan, "A", "B", np.nan, "A"],
"Treatment": ["Basic", np.nan, "Advanced", "Basic",
"Advanced", "Basic", np.nan, "Basic"]
})

numerical_cols = ["Patient_Age", "Blood_Sugar"]
categorical_cols = ["Blood_Group", "Treatment"]

constant_preprocessor = ColumnTransformer(
transformers=[
(
"numerical",
SimpleImputer(strategy="constant", fill_value=0),
numerical_cols
),
(
"categorical",
SimpleImputer(strategy="constant", fill_value="Unknown"),
categorical_cols
)
],
verbose_feature_names_out=False
)

result = constant_preprocessor.fit_transform(df)

df_constant = pd.DataFrame(
result,
columns=constant_preprocessor.get_feature_names_out()
)

df_constant = df_constant[df.columns]

print("Dataset After Constant Imputation:")
print(df_constant)

print("\nMissing Values:")
print(df_constant.isnull().sum())

print("\nAny Missing Values?")
print(df_constant.isnull().values.any())
```

```python
#Question 3: Investigating the Impact of Outliers on Imputation
#A. Impute missing numerical values using both mean and median strategies with SimpleImputer. Compare the imputed values and calculate the mean and standard deviation of Annual_Spending after each method. Explain the effect of the unusually large spending value on the mean.
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

df = pd.DataFrame({
"Customer_ID": range(1, 11),
"Annual_Spending": [12000, 15000, np.nan, 18000, 22000,
25000, 30000, np.nan, 900000, 20000],
"Order_Count": [4, 5, 6, np.nan, 8, 7, 9, 10, 50, np.nan],
"Membership": ["Silver", "Gold", "Silver", "Gold", np.nan,
"Silver", "Gold", np.nan, "Platinum", "Silver"]
})

numerical_cols = ["Annual_Spending", "Order_Count"]

# Mean imputation

mean_imputer = SimpleImputer(strategy="mean")
df_mean = df.copy()

df_mean[numerical_cols] = mean_imputer.fit_transform(
df[numerical_cols]
)

# Median imputation

median_imputer = SimpleImputer(strategy="median")
df_median = df.copy()

df_median[numerical_cols] = median_imputer.fit_transform(
df[numerical_cols]
)

print("Mean Imputation Values:")
print(pd.Series(mean_imputer.statistics_, index=numerical_cols))

print("\nMedian Imputation Values:")
print(pd.Series(median_imputer.statistics_, index=numerical_cols))

print("\nMean-Imputed Annual Spending:")
print(df_mean["Annual_Spending"])

print("\nMedian-Imputed Annual Spending:")
print(df_median["Annual_Spending"])

# Compare annual spending statistics

comparison = pd.DataFrame({
"Original_Available_Data": df["Annual_Spending"].agg(
["mean", "median", "std"]
),
"Mean_Imputation": df_mean["Annual_Spending"].agg(
["mean", "median", "std"]
),
"Median_Imputation": df_median["Annual_Spending"].agg(
["mean", "median", "std"]
)
})

print("\nAnnual Spending Statistics:")
print(comparison.round(2))
```

```python
#B. Build a reusable preprocessing pipeline using ColumnTransformer and SimpleImputer that applies median imputation to numerical features and most-frequent imputation to categorical features. Exclude Customer_ID from the input features, then verify that the transformed dataset contains no missing values and retains the expected feature columns
import pandas as pd
import numpy as np

from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

df = pd.DataFrame({
"Customer_ID": range(1, 11),
"Annual_Spending": [12000, 15000, np.nan, 18000, 22000,
25000, 30000, np.nan, 900000, 20000],
"Order_Count": [4, 5, 6, np.nan, 8, 7, 9, 10, 50, np.nan],
"Membership": ["Silver", "Gold", "Silver", "Gold", np.nan,
"Silver", "Gold", np.nan, "Platinum", "Silver"]
})

# Exclude Customer_ID

X = df[["Annual_Spending", "Order_Count", "Membership"]]

numerical_cols = ["Annual_Spending", "Order_Count"]
categorical_cols = ["Membership"]

# Create the preprocessing transformer

preprocessor = ColumnTransformer(
transformers=[
(
"numerical",
SimpleImputer(strategy="median"),
numerical_cols
),
(
"categorical",
SimpleImputer(strategy="most_frequent"),
categorical_cols
)
],
verbose_feature_names_out=False
)

# Create a reusable pipeline

pipeline = Pipeline([
("preprocessing", preprocessor)
])

# Fit and transform the input data

transformed = pipeline.fit_transform(X)

# Restore column names

feature_names = pipeline.named_steps[
"preprocessing"
].get_feature_names_out()

df_transformed = pd.DataFrame(
transformed,
columns=feature_names
)

# Restore numerical data types

for column in numerical_cols:
    df_transformed[column] = pd.to_numeric(
        df_transformed[column]
    )

print("Transformed Dataset:")
print(df_transformed)

print("\nFeature Columns:")
print(df_transformed.columns.tolist())

print("\nMissing Values:")
print(df_transformed.isnull().sum())

print("\nAny Missing Values?")
print(df_transformed.isnull().values.any())

print("\nCustomer_ID Included?")
print("Customer_ID" in df_transformed.columns)
```
