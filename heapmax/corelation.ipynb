```python
df1=sns.load_dataset('tips')[['total_bill','tip','time']]
```

```python
df1
```

```python
corr_matrix = df1.corr(numeric_only=True)
print(corr_matrix)
```

```python
# Correlation between Experience and Salary
exp_sal_corr = df1["sex"].corr(df1["Salary"])
print(f"Correlation: {exp_sal_corr:.2f}")
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(5,3))
sns.heatmap(annot=True)
```

```python
df1.corr(numeric_only=True)["tip"].sort_values(ascending=False)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
sns.heatmap(df1.corr(numeric_only=True), annot=True, cmap="coolwarm", fmt=".2f", vmin=-1, vmax=1)
plt.title("Correlation Matrix Heatmap")
plt.show()
```
