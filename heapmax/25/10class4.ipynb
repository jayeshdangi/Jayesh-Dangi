```python
from sklearn.preprocessing import PowerTransformer
import numpy as np
```

```python
X=np.array([
    [-1],
    [0],
    [3],
    [5],
    [10],
    [50],
])
```

```python
pt=PowerTransformer(method="yeo-johnson")
```

```python
X_transformed=pt.fit_transform(X)
```

```python
print("Original:")
print(X.ravel())
```

```python
print("\nTransformed:")
print(np.round(X_transformed.ravel(),3))
print("\nLearned lambda:")
print(pt.lambdas_)
```

```python
import numpy as np
from sklearn.preprocessing import PowerTransformer

# Right-skewed feature (2D array shaped 5x1)
y = np.array([[1], [2], [3], [10], [50]])

# Method must be "box-cox" with a hyphen (-)
pt = PowerTransformer(method="box-cox")
y_transformed = pt.fit_transform(y)

print('Original:')
print(y.ravel())

print('\nTransformed:')
print(np.round(y_transformed.ravel(), 3))

print('\nLearned lambda:')
print(pt.lambdas_)
```

```python
import pandas as pd
import numpy as np

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Henry'],
    # Numerical features with NaN
    'Age': [25, np.nan, 30, 35, np.nan, 40, 28, 32],
    'Salary': [50000, 60000, np.nan, 80000, 55000, np.nan, 62000, 75000],
    # Categorical features with NaN
    'City': ['New York', 'London', np.nan, 'Paris', 'London', 'New York', np.nan, 'Paris'],
    'Department': ['HR', 'IT', 'IT', np.nan, 'Sales', 'HR', 'Sales', np.nan],
    # Target / Feature
    'Experience_Years': [2, 5, 7, 10, 3, 12, np.nan, 8]
}

df = pd.DataFrame(data)
df
```

```python
from sklearn.impute import SimpleImputer
```

```python
imputer=SimpleImputer(strategy='mean')
imputer.fit_transform(df[['Salary']])
```

```python
imputer=SimpleImputer(strategy='mean')
imputer.fit_transform(df[['Salary']]).flatten()
```

```python
imputer=SimpleImputer(strategy='mean')

df['new_salary']=imputer.fit_transform(df[['Salary']]).flatten()
```

```python
df
```

```python
imputer=SimpleImputer(strategy='median')
imputer.fit_transform(df[['Salary']]).flatten()
```

```python
imputer=SimpleImputer(strategy='median')

df['new_Experience_Years']=imputer.fit_transform(df[['Experience_Years']]).flatten()
```

```python
df
```

```python
imputer=SimpleImputer(strategy='most_frequent')

df['new_Department']=imputer.fit_transform(df[['Department']]).flatten()
```
